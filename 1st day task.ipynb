```python
df.drop(columns=['date_added'], inplace=True)
```

```python
df['country'].fillna('Unknown', inplace=True)
df
```

```python
df.isnull().sum()
```

```python
df['director'].fillna('Unknown', inplace=True)
df['cast'].fillna('Unknown', inplace=True)
df
```

```python
df['rating'].fillna('NR', inplace=True)
df
```

```python
df.isnull().sum()
```

```python
# Strip whitespace from string columns
text_cols = ['title', 'director', 'cast', 'country', 'rating', 'listed_in', 'description']
for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
# Extract duration numerical values
df['duration_num'] = df['duration'].str.extract('(\d+)').astype(float)

# Separate unit type (min vs Season/Seasons)
df['duration_unit'] = df['duration'].str.extract('([a-zA-Z]+)')
```

```python
# Create a list of genres for each row
df['genres'] = df['listed_in'].str.split(', ')

# Example: Get primary genre
df['primary_genre'] = df['genres'].str[0]
```

```python
# Convert type and rating to category
df['type'] = df['type'].astype('category')
df['rating'] = df['rating'].astype('category')
```

```python
# Save cleaned dataframe
df.to_csv('amazon_prime_titles_cleaned.csv', index=False)
```
